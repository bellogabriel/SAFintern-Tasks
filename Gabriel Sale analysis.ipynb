```python
# Import necessary libraries
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Load the dataset
com_sales_df = pd.read_csv('company_sales_data.csv')
com_sales_df
```

```python
# Inspect the data
com_sales_df.info()
```

```python
# Convert the month number to month names
com_sales_df['month_name'] = com_sales_df['month_number'].apply(lambda x: pd.to_datetime(x, format='%m').strftime('%B'))
com_sales_df
```

```python
# Exercise 1: Read Total profit of all months and show it using a line plot.
plt.figure(figsize=(8, 5))
plt.plot(com_sales_df['month_name'], com_sales_df['total_profit'], marker='o', color='green')
plt.title('Total Profit by Months', fontsize=14, fontweight="bold", pad=15)
plt.xlabel('Month', fontsize=10)
plt.ylabel('Total Profit', fontsize=10)
plt.xticks(com_sales_df['month_name'], rotation=45, fontsize=8)
plt.yticks(fontsize=8)
plt.grid(True, color="gray", linewidth=0.15)
plt.show()
```

#### INSIGHT FROM PLOT
1. Q1 and Q2 both show relatively lower total sales averaging a little of 200,000 per month for this period.
2. The total sales value then begins to rise during the summer period, particularly in the months of July and August.
3. We observe a dip in the values during the months of September and October before rising again for the last two months.
4. This shows that sales then to boom during the summer holiday and the end-of-year seasons. Groceries items tend to sell better during these periods because many customers purchase more items because of the travels.

```python
# Exercise 2: Read data for Bathing soap and facewash of all months and display it using the Subplot
plt.figure(figsize=(12, 6))
plt.subplot(1, 2, 1)
plt.plot(com_sales_df['month_name'], com_sales_df['bathingsoap'], marker='o', color='blue')
plt.title('Bathing Soap Sales by Months', fontsize=14, fontweight="bold", pad=15)
plt.xlabel('Month', fontsize=10)
plt.ylabel('Sales', fontsize=10)
plt.xticks(com_sales_df['month_name'], rotation=45, fontsize=8)
plt.yticks(fontsize=8)
plt.grid(True, color="gray", linewidth=0.15)
plt.subplot(1, 2, 2)
plt.plot(com_sales_df['month_name'], com_sales_df['facewash'], marker='o', color='orange')
plt.title('Facewash Sales by Months', fontsize=14, fontweight="bold", pad=15)
plt.xlabel('Month', fontsize=10)
plt.ylabel('Sales', fontsize=10)
plt.xticks(com_sales_df['month_name'], rotation=45, fontsize=8)
plt.yticks(fontsize=8)
plt.grid(True, color="gray", linewidth=0.15)
plt.show()
```

#### Insight from Plot

1. Comparison between similar grocery items show a close relation in sales pattern with some significant differences.
2. There is generally higher sales towards the end of the year for both commodities.
3. January seems to be amoung the highest performing months among the first two quarters (Q1 and Q2).
4. The sale of facewash shows a severe dip in the month of July recording its lowest sales in that month.
5. While the sale of bathing soap has its lowest sale in the month of february.
6. The month of August to December should be targetting for promotional campaigns and high inventory restocking period. This will help the marketer to take advantage of these relatively higher sales period and maximise profit.
